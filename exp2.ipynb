```python
import numpy as np
from sklearn.datasets import load_breast_cancer
```

```python
data = load_breast_cancer()
X = data.data
y = data.target
X = X.T
y = y.reshape(1,-1)
```

```python
print(X.shape)
```

```python
# we gonna have 30  inputs, 16 L1 , 8L2 ,1L3
def intialize_parameters():
    w1 = np.random.randn(16,30)*0.01
    b1 = np.zeros((16,1))

    w2 = np.random.randn(8,16)*0.01
    b2 = np.zeros((8,1))

    w3 = np.random.randn(1,8)*0.01
    b3 = np.zeros((1,1))

    return w1,b1,w2,b2,w3,b3
```

```python
def ReLU(z):
    return np.maximum(0,z)

def sigmoid(z):
    return 1/(1+np.exp(-z))

def ReLU_derivative(z):
    return(z>0).astype(float) #it basically uses some boolean matrix here
```

```python
def forward(w1,b1,w2,b2,w3,b3,X):
    z1 = np.dot(w1,X)+b1
    a1 = ReLU(z1)

    z2 = np.dot(w2,a1)+b2
    a2 = ReLU(z2)

    z3 = np.dot(w3,a2) + b3
    a3 = sigmoid(z3)

    return z1,a1,z2,a2,z3,a3
```

```python
def cost(a3,y):
    m = X.shape[1]

    value = -(1/m)*np.sum(y*np.log(a3)+(1-a3)*(np.log(1-a3)))

    return value
```

```python
def backward(z1,a1,z2,a2,a3,w2,w3):
    m = X.shape[1]
    dz3 = a3 - y
    dw3 = (1/m)*(np.dot(dz3,a2.T))
    db3 = (1/m)*np.sum(dz3,axis=1,keepdims=True)

    dz2 = (np.dot(w3.T,dz3))*(ReLU_derivative(z2))
    dw2 = (1/m)*(np.dot(dz2,a1.T))
    db2 = (1/m)*(np.sum(dz2,axis=1,keepdims=True))

    dz1 = (np.dot(w2.T,dz2))*(ReLU_derivative(z1))
    dw1 = (1/m)*(np.dot(dz1,X.T))
    db1 = (1/m)*(np.sum(dz1,axis=1,keepdims=True))

    return dz3,dw3,db3,dz2,dw2,db2,dz1,dw1,db1
```

```python
def learning(dw3,db3,dw2,db2,dw1,db1,w2,w3,learning_rate,w1):
    w1 = w1 - learning_rate*(dw1)
    b1 = b1 - learning_rate*(db1)

    w2 = w2 - learning_rate*(dw2)
    b2 = b2 - learning_rate*(db2)

    w3 = w3 - learning_rate*(dw3)
    b3 = b3 - learning_rate*(db3)

    return w1,b1,w2,b2,w3,b3
```

```python
w1,b1,w2,b2,w3,b3 = intialize_parameters()

for i in range(10000):
    z1,a1,z2,a2,z3,a3 = forward(w1,b1,w2,b2,w3,b3,X)

    m = X.shape[1]

    J = cost(a3,y)


    dz3,dw3,db3,dz2,dw2,db2,dz1,dw1,db1 = backward(z1,a1,z2,a2,a3,w2,w3)
    learning_rate = 0.01
    print(w1.shape)
    print(dw1.shape)
    w1,b1,w2,b2,w3,b3 = learning(dw3,db3,dw2,db2,dw1,db1,w2,w3,learning_rate,w1)


    if i%1000 == 0:
        print(i,J)
    
```
